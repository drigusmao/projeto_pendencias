# Tratamento de Dados - Camada Silver

Este notebook processa e limpa os dados da camada Bronze para a camada Silver:
* **Validação de dados**: identifica e trata valores nulos e inconsistências
* **Padronização**: garante tipos de dados corretos e formatos consistentes
* **Limpeza**: remove duplicatas e corrige erros de entrada

Tabelas processadas:
* `silver_agencias` - Agências validadas
* `silver_clientes` - Clientes com dados tratados
* `silver_contratos` - Contratos validados com relacionamentos corretos

```python
%python
# Importar bibliotecas necessárias
from pyspark.sql import SparkSession
from pyspark.sql.types import *

# Inicializar SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
%python
# Ler tabela bronze de agências
df_agencias = spark.read.table("workspace.default.bronze_agencias")

# Contagem de registros, visualizar dados e esquema
print(f"Total de registros: {df_agencias.count()}")
display(df_agencias)

# Verificar campos nulos e se houver algum contar por coluna 
for col in df_agencias.columns:
    null_count = df_agencias.filter(df_agencias[col].isNull()).count()
    if null_count > 0:
        print(f"Coluna '{col}': {null_count} nulos")

# Criar tabela Silver 
df_agencias_silver = df_agencias

# Salvar como tabela Silver
df_agencias_silver.write.mode("overwrite").saveAsTable("workspace.default.silver_agencias")
```

```python
# Ler tabela bronze de clientes
df_clientes = spark.read.table("workspace.default.bronze_clientes")

# Contagem de registros, visualizar dados e esquema
print(f"Total de registros: {df_clientes.count()}")
display(df_clientes)

# Verificar campos nulos e se houver algum contar por coluna 
for col in df_clientes.columns:
    null_count = df_clientes.filter(df_clientes[col].isNull()).count()
    if null_count > 0:
        print(f"Coluna '{col}': {null_count} nulos")

# Criar tabela Silver 
df_clientes_silver = df_clientes

# Salvar como tabela Silver
df_clientes_silver.write.mode("overwrite").saveAsTable("workspace.default.silver_clientes")
```
